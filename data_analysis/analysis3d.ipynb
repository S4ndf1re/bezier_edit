```python
import open3d as o3d
from open3d import core as o3c
import json
```

```python
# with open("../click_log/right/1754043015998.json") as f:
# with open("../click_log/right/1754043020139.json") as f:
with open("../click_log/right/1754043017700.json") as f:
    data = json.load(f)
print(data)
```

```python
transforms = data["timed_transforms"]

transforms = list(map(lambda t: t[1]["translation"], transforms))
normals = []
lines = []
for i in range(len(transforms)-1):
    a = transforms[i]
    b = transforms[i+1]
    diff = [b[0]- a[0], b[1]- a[1], b[2]- a[2]]
    normals.append(diff)
    lines.append([i, i+1])
normals.append([0,0,0])

derivatives = data["controller_derivatives"]
colors = list(map(lambda t: [t[1], 0, 0], derivatives))
print(transforms)
print(normals)
```

```python
pcd = o3d.t.geometry.PointCloud(transforms)
lines = o3d.t.geometry.LineSet(transforms, lines)
pcd.point.colors = colors
pcd.point.normals = normals
legacy = pcd.to_legacy()
o3d.visualization.draw_geometries([legacy, lines.to_legacy()], point_show_normal=True)
```
